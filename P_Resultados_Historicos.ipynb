```python
import pandas as pd
import numpy as np

# Cargar el archivo
df_res = pd.read_csv('resultados_historicos.csv')

# 1. Estandarizar textos clave
df_res['tipo_eleccion'] = df_res['tipo_eleccion'].str.strip().str.lower()
df_res['partido'] = df_res['partido'].str.strip().str.upper()
df_res['candidato'] = df_res['candidato'].str.strip()

# 2. Manejar valores nulos en 'votos_validos' (rellenar con 0 si faltan)
df_res['votos_validos'] = df_res['votos_validos'].fillna(0)

# 3. Asegurar que las columnas numéricas sean enteros o floats correctos
cols_numericas = ['votos_validos', 'votos_nulos', 'votos_en_blanco', 'total_empadronados_mesa', 'total_votantes_mesa']
for col in cols_numericas:
    df_res[col] = pd.to_numeric(df_res[col], errors='coerce').fillna(0).astype(int)

# 4. Eliminar filas duplicadas si las hubiera
df_res = df_res.drop_duplicates()

# 5. Guardar el resultado limpio
df_res.to_csv('resultados_historicos_limpio.csv', index=False)
print("¡resultados_historicos.csv limpiado y guardado con éxito!")
print(f"Dimensiones finales: {df_res.shape}")
```
