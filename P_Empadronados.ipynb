```python
import pandas as pd
import numpy as np

# 1. Cargar los archivos directamente (como ya están en la misma carpeta en Jupyter)
df_2 = pd.read_csv('empadronados_20260302.csv')
df_3 = pd.read_csv('empadronados_20260303.csv')
df_4 = pd.read_csv('empadronados_20260304.csv')

# 2. Combinar los dataframes en uno solo
df = pd.concat([df_2, df_3, df_4], ignore_index=True)

print(f"Dimensiones del dataset combinado: {df.shape}")
df.head()

# Ver valores nulos por columna
print("--- Valores Nulos ---")
print(df.isnull().sum())

# Revisar valores únicos antes de limpiar
print("\n--- Valores únicos en 'sexo' ---")
print(df['sexo'].unique())

print("\n--- Valores únicos en 'estado_registro' ---")
print(df['estado_registro'].unique())


# 1. Limpiar y unificar la columna 'sexo'
# Convertimos a minúsculas y mapeamos a un formato estándar ('M' y 'F')
df['sexo'] = df['sexo'].str.strip().str.lower()
sexo_map = {
    'm': 'M', 'masculino': 'M',
    'f': 'F', 'femenino': 'F'
}
df['sexo'] = df['sexo'].map(sexo_map).fillna('DESCONOCIDO')

# 2. Limpiar y unificar la columna 'estado_registro'
# Uniformizamos a minúsculas y removemos espacios
df['estado_registro'] = df['estado_registro'].str.strip().str.lower()

# 3. Uniformizar los formatos de 'rango_edad'
df['rango_edad'] = df['rango_edad'].str.strip()
edad_map = {
    '26 a 35': '26-35',
    '18 a 25': '18-25',
    'mayor de 60': '61+'
}
df['rango_edad'] = df['rango_edad'].replace(edad_map)

# 4. Convertir columnas de fecha al tipo datetime de Pandas
df['fecha_inscripcion'] = pd.to_datetime(df['fecha_inscripcion'], errors='coerce')
df['fecha_ultima_actualizacion'] = pd.to_datetime(df['fecha_ultima_actualizacion'], errors='coerce')

# 5. Eliminar filas duplicadas si las hubiera (basado en el ID del empadronado)
df = df.drop_duplicates(subset=['id_empadronado'], keep='last')

print("¡Limpieza aplicada con éxito!")
print(f"Nuevas dimensiones del dataset: {df.shape}")

# Comprobamos cómo quedaron los valores únicos después de limpiar
print("--- Estado de 'sexo' limpio ---")
print(df['sexo'].value_counts())

print("\n--- Estado de 'estado_registro' limpio ---")
print(df['estado_registro'].value_counts())

# Guardar el resultado limpio en un nuevo archivo CSV
df.to_csv('empadronados_limpios_total.csv', index=False)
print("\nArchivo guardado exitosamente como 'empadronados_limpios_total.csv'")
```
