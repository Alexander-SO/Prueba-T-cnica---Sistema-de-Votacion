```python
import pandas as pd
import numpy as np

# Cargar el archivo
df_cv = pd.read_csv('centros_votacion.csv')

# 1. Estandarizar textos de departamento y municipio (quitar espacios y normalizar mayúsculas)
df_cv['nombre_departamento'] = df_cv['nombre_departamento'].str.strip().str.title()
df_cv['nombre_municipio'] = df_cv['nombre_municipio'].str.strip().str.title()
df_cv['nombre_centro'] = df_cv['nombre_centro'].str.strip()

# 2. Corregir la columna 'area' (unificar 'U'/'urbana' y 'R'/'rural')
df_cv['area'] = df_cv['area'].str.strip().str.lower()
area_map = {'u': 'urbana', 'urbana': 'urbana', 'r': 'rural', 'rural': 'rural'}
df_cv['area'] = df_cv['area'].map(area_map).fillna('desconocido')

# 3. Corregir la columna 'latitud' (reemplazar comas por puntos si existen y convertir a float)
if df_cv['latitud'].dtype == 'object':
    df_cv['latitud'] = df_cv['latitud'].astype(str).str.replace(',', '.')
df_cv['latitud'] = pd.to_numeric(df_cv['latitud'], errors='coerce')

# 4. Rellenar valores nulos en 'zona' si es necesario
df_cv['zona'] = df_cv['zona'].fillna('Sin Zona')

# 5. Guardar el resultado limpio
df_cv.to_csv('centros_votacion_limpio.csv', index=False)
print("¡centros_votacion.csv limpiado y guardado con éxito!")
print(f"Dimensiones finales: {df_cv.shape}")
```
